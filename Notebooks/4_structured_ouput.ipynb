```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model_name="gpt-6.1-sol")
```

```python
text1="Hello My name is sarfraz", \
"My email is sarfrazsha@gmail.com"
```

```python
res=llm.invoke(f"Just tell my name and email from: {text1}")
res.text
```

```python
from pydantic import BaseModel, Field

class ResponseStructure(BaseModel):
    name: str = Field(description="Name of the person")
    email: str = Field(description="Email of the person")

structured_llm = llm.with_structured_output(ResponseStructure)
result = structured_llm.invoke(f"Just tell my name and email from: {text1}")
result
result.name
```
