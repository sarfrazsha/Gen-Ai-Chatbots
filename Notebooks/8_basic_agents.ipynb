```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(num1:int, num2:int):
    """This tool will return the sum of two numbers."""
    return num1+num2
```

```python
add_numbers.invoke({"num1":12,"num2":3})
```

```python
from langchain.agents import create_agent
```

```python

from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model_name="gpt-6.1-sol");
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers],
    system_prompt="You are a helpful assistant that can add numbers using the add_numbers tool."
)
```

```python
response=agent.invoke({"messages":[{"role":"user","content":"please add 12 and 3"}]})
```
