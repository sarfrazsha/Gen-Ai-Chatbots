```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm=ChatGroq(model="openai/gpt-oss-safeguard-20b")
search=GoogleSerperAPIWrapper()
```

```python
agent=create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are an agent and you have to access the internet to answer the questions. You have access to a tool called 'search' which you can use to search the internet for information. Use this tool to find the information you need to answer the user's question. If you don't know the answer, use the search tool to find it.",

    checkpointer=MemorySaver()
)
```

```python
question="Total Runs?"
res=agent.invoke({"messages":[{"role":"user","content":question}]},
                 {"configurable":{"thread_id":"1"}})
```
